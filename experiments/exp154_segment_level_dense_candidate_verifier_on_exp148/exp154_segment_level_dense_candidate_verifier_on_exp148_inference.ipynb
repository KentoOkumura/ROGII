# exp154_segment_level_dense_candidate_verifier_on_exp148 inference

exp148 saved-booster base prediction に、train-side best の segment-level dense verifier を適用して `submission.csv` を生成する。

## 1. Setup and selected contract

```python
from __future__ import annotations

import json

from settings import EXPERIMENT_NAME, ExperimentPaths, get_nested
from segment_level_dense_candidate_verifier_on_exp148 import run_inference_from_config

def cfg_get(config, dotted_key, default=None):
    value = get_nested(config, dotted_key)
    return default if value is None else value

paths = ExperimentPaths()
paths.require_kaggle_runtime()
paths.ensure_output_dirs()
config = paths.config

print('Experiment:', EXPERIMENT_NAME)
print('Route:', cfg_get(config, 'experiment.route'))
print('Inference mode:', cfg_get(config, 'inference.mode'))
print('Selected verifier:', cfg_get(config, 'inference.selected_variant'))
print('Exp148 base model:', cfg_get(config, 'inference.exp148.selected_model'))
print('Output dir:', paths.artifacts_dir)
print('Submission path:', paths.submission_path)
```

## 2. Generate exp148 base, apply verifier, and write submission

```python
summary = run_inference_from_config(
    config,
    output_dir=paths.artifacts_dir,
    submission_path=paths.submission_path,
    sample_submission_path=paths.sample_submission_path,
    raw_data_dir=paths.raw_data_dir,
    test_dir=paths.test_data_dir,
)
paths.metrics_path.write_text(
    json.dumps(summary, indent=2, ensure_ascii=False, sort_keys=True) + '\n'
)
print(json.dumps(summary['metrics'], indent=2, ensure_ascii=False))
print('Submission:', paths.submission_path)
```
