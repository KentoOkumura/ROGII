## 1. Setup and configuration


```python
from __future__ import annotations

import json
from pathlib import Path

from cv_submit_gate import main as run_gate
from settings import EXPERIMENT_NAME, ExperimentPaths, get_nested, load_config

paths = ExperimentPaths()
paths.ensure_output_dirs()
config = load_config()

print('Experiment:', EXPERIMENT_NAME)
print('Route:', get_nested(config, 'experiment.route'))
print('Parent:', get_nested(config, 'lineage.parent'))
print('Source metrics:', get_nested(config, 'data.source_metrics'))
print('Source submission:', get_nested(config, 'data.source_submission'))
print('Overlap probe:', get_nested(config, 'data.overlap_probe_metrics'))
print('Artifacts:', paths.artifacts_dir)
```

## 2. Source evidence


```python
source_metrics_path = paths.root / get_nested(config, 'data.source_metrics')
probe_metrics_path = paths.root / get_nested(config, 'data.overlap_probe_metrics')

source_metrics = json.loads(source_metrics_path.read_text())
probe_metrics = json.loads(probe_metrics_path.read_text())

print('Source status:', source_metrics.get('status'))
print('Source CV:', source_metrics.get('inference_candidate', {}).get('rmse_tvt'))
print('Source inference status:', source_metrics.get('inference', {}).get('status'))
print('Source submit-check:', source_metrics.get('inference', {}).get('submit_check', {}).get('status'))
print('Probe status:', probe_metrics.get('hidden_code_submission', {}).get('status'))
print('Probe interpretation:', probe_metrics.get('hidden_code_submission', {}).get('interpretation'))
```

## 3. Gate execution


```python
run_gate([])
```

## 4. Metrics and artifacts


```python
metrics = json.loads(paths.metrics_path.read_text())
decision = json.loads((paths.artifacts_dir / 'cv_submit_gate_decision.json').read_text())

print('Decision:', decision['decision'])
print('Approved:', decision['approved_for_code_submit'])
print('Required rules:', metrics['required_rule_pass_count'], '/', metrics['required_rule_count'])
print('Submit command:')
print(decision['submit_target']['command'])
print('Artifacts:')
for name in ['cv_submit_gate_decision.json', 'cv_submit_gate_decision.csv', 'cv_submit_gate_report.md']:
    print('-', paths.artifacts_dir / name, Path(paths.artifacts_dir / name).exists())
```
