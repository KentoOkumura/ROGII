# exp066_cv_submit_gate inference

This experiment does not create a new submission. It records the submit target from the gate report.

## Contents

1. Setup and paths
2. Gate report
3. No new submission


## 1. Setup and paths


```python
from __future__ import annotations

import json

from cv_submit_gate import main as run_gate
from settings import EXPERIMENT_NAME, ExperimentPaths, get_nested, load_config

paths = ExperimentPaths()
paths.ensure_output_dirs()
config = load_config()

print('Experiment:', EXPERIMENT_NAME)
print('Route:', get_nested(config, 'experiment.route'))
print('Submit source kernel:', get_nested(config, 'model.candidate.inference_kernel'))
print('Submit source version:', get_nested(config, 'model.candidate.inference_kernel_version'))
```

## 2. Gate report


```python
if not (paths.artifacts_dir / 'cv_submit_gate_decision.json').exists():
    run_gate([])

decision = json.loads((paths.artifacts_dir / 'cv_submit_gate_decision.json').read_text())
print('Decision:', decision['decision'])
print('Approved:', decision['approved_for_code_submit'])
print('Source submission:', decision['submit_target']['source_submission_path'])
print('Submit command:')
print(decision['submit_target']['command'])
```

## 3. No new submission


```python
no_submission = paths.artifacts_dir / 'NO_SUBMISSION.txt'
no_submission.write_text(
    'exp066_cv_submit_gate does not create submission.csv. Submit exp063 inference v2 if the gate is accepted.\n'
)
print(no_submission)
```
