## 1. Setup and guardrails


```python
from __future__ import annotations

import json
from pathlib import Path

import pandas as pd

from settings import EXPERIMENT_NAME, ExperimentPaths, load_config

paths = ExperimentPaths()
paths.require_kaggle_runtime()
paths.ensure_output_dirs()
config = load_config()

print('Experiment:', EXPERIMENT_NAME)
print('Route:', config['experiment']['route'])
print('GPU enabled:', config['runtime']['kaggle']['enable_gpu'])
print('Submission path reserved by template:', paths.submission_path)
print('This notebook does not create a submission file.')
```

## 2. Optional audit rerun


```python
from tabicl_artifact_diversity_audit import run_audit

metrics = run_audit()
print(json.dumps(metrics, indent=2, sort_keys=True))

if paths.submission_path.exists():
    raise RuntimeError(f'Unexpected submission file was created: {paths.submission_path}')
```

## 3. Output summary


```python
summary_path = Path(metrics['artifacts']['summary'])
readme_path = Path(metrics['artifacts']['readme'])
print('Summary:', summary_path)
print('README:', readme_path)
print(readme_path.read_text()[:4000])
```
