## 1. Setup and paths


```python
from __future__ import annotations

import os
import shutil

from settings import EXPERIMENT_NAME, ExperimentPaths

DEBUG = os.environ.get("EXPERIMENT_DEBUG", "0") == "1"

paths = ExperimentPaths()
paths.require_kaggle_runtime()
paths.ensure_output_dirs()

print("Experiment:", EXPERIMENT_NAME)
print("Root:", paths.root)
print("Test data:", paths.test_data_dir)
print("Sample submission:", paths.sample_submission_path)
print("Submission path:", paths.submission_path)
print("Debug:", DEBUG)
```

## 2. Submission creation


```python
sample_submission = paths.sample_submission_path
submission_path = paths.submission_path

if not sample_submission.exists():
    raise FileNotFoundError(f"sample submission not found: {sample_submission}")

shutil.copyfile(sample_submission, submission_path)
print("Created Kaggle submission:", submission_path)
```
