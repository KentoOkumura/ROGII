## 1. Setup and configuration

```python
from pathlib import Path

import pandas as pd

from public_sel15_pf_oof import run_feature_generation
from settings import EXPERIMENT_NAME, ExperimentPaths, get_nested, load_config

paths = ExperimentPaths()
paths.require_kaggle_runtime()
paths.ensure_output_dirs()
config = load_config()

print(f"Experiment: {EXPERIMENT_NAME}")
print(f"Route: {get_nested(config, 'experiment.route')}")
print(f"Parent: {get_nested(config, 'lineage.parent')}")
print(f"Train data: {paths.train_data_dir}")
print(f"Features dir: {paths.features_dir}")
print(f"Cutoffs: {get_nested(config, 'model.cutoff_fractions')}")
print(f"PF particles: {get_nested(config, 'model.public_sel15_pf.n_particles')}")
print(f"PF seeds: {get_nested(config, 'model.public_sel15_pf.n_seeds')}")
print(f"Debug wells: {get_nested(config, 'runtime.debug_n_wells')}")
```

## 2. Generate PF/Beam OOF-like features

```python
metrics = run_feature_generation(paths, config)
metrics
```

## 3. Metrics and artifacts

```python
feature_path = Path(metrics['feature_file'])
well_summary_path = Path(metrics['well_summary_file'])

print(f"Feature file: {feature_path}")
print(f"Well summary: {well_summary_path}")
print(f"Rows: {metrics['rows']}")
print(f"PF RMSE diagnostic: {metrics['pf_rmse']}")
print(f"Last-anchor RMSE diagnostic: {metrics['last_anchor_rmse']}")
print(f"Beam RMSE diagnostic: {metrics['beam_rmse']}")

display(pd.read_csv(well_summary_path).head())
display(pd.read_csv(feature_path, nrows=5))
```
