# exp085_u_projection_feature_ablation inference

Guard notebook. Inference is intentionally disabled until the train-side feature ablation selects a variant and parity checks are documented.

## 1. Setup and configuration

```python
from settings import load_config, get_nested

def cfg_get(config, dotted_key, default=None):
    value = get_nested(config, dotted_key)
    return default if value is None else value

config = load_config()
print("Experiment:", config["experiment"]["name"])
print("Inference mode:", cfg_get(config, "inference.mode"))
print("Selected variant:", cfg_get(config, "inference.selected_variant"))
print("Selected model:", cfg_get(config, "inference.selected_model"))
```

## 2. Selection guard

```python
selected_variant = cfg_get(config, "inference.selected_variant")
if not selected_variant:
    raise RuntimeError(
        "No selected U-projection feature variant. Run train-side ablation, review CV/parity, "
        "then set inference.selected_variant before building a submission notebook."
    )
```
