# exp125_confidence_gate_continuity_rawtest_parity inference

Diagnostic-only inference notebook. It writes a no-submission summary and intentionally does not create `submission.csv`.

## Contents

1. Setup and paths
2. No-submission contract

## 1. Setup and paths

```python
from __future__ import annotations

import json
from datetime import UTC, datetime

from settings import EXPERIMENT_NAME, ExperimentPaths, get_nested, load_config

paths = ExperimentPaths()
paths.require_kaggle_runtime()
paths.ensure_output_dirs()
config = load_config()

print("Experiment:", EXPERIMENT_NAME)
print("Inference mode:", get_nested(config, "inference.mode"))
print("Artifacts:", paths.artifacts_dir)
print("Submission path would be:", paths.submission_path)
```

## 2. No-submission contract

```python
summary = {
    "experiment": EXPERIMENT_NAME,
    "status": "inference_disabled_diagnostic_only",
    "created_at": datetime.now(UTC).isoformat(),
    "selected_variant": get_nested(config, "inference.selected_variant"),
    "submission_created": False,
    "reason": "exp125 is a train-side continuity/raw-test parity audit. Implement an explicit inference port only after a gate passes the train-side guardrails.",
}
out = paths.artifacts_dir / "exp125_confidence_gate_continuity_rawtest_parity_inference_disabled_summary.json"
out.write_text(json.dumps(summary, indent=2, ensure_ascii=False, sort_keys=True) + "\n")
print(json.dumps(summary, indent=2, ensure_ascii=False))
print("Wrote:", out)
```
