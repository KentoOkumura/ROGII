```python
import pandas as pd
import matplotlib.pyplot as plt

# データの場所を設定
base_path = '/kaggle/input/competitions/rogii-wellbore-geology-prediction'

# trainフォルダにある「000d7d20」という井戸の基準井（Typewell）データを読み込む
typewell_df = pd.read_csv(f'{base_path}/train/000d7d20__typewell.csv')

# データの中身を表示
print("【基準井戸（Typewell）のデータ】")
display(typewell_df.head())

# 地層の深さ(TVT)に対するガンマ線(GR)の基本波形をグラフ化
plt.figure(figsize=(12, 4))
plt.plot(typewell_df['TVT'], typewell_df['GR'], color='green', linewidth=1.5)
plt.title('Typewell Blueprint: Gamma Ray (GR) by True Vertical Thickness (TVT)')
plt.xlabel('True Vertical Thickness [TVT] (地層の深さ)')
plt.ylabel('Gamma Ray [GR] (ガンマ線)')
plt.grid(True)
plt.show()
```
