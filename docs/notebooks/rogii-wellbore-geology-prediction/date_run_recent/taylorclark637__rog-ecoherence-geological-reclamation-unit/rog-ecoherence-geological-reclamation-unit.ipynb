```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import json
import pickle
from datetime import datetime
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import GradientBoostingRegressor, RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import warnings
warnings.filterwarnings('ignore')

# ============================================================================
# USE ROOT DIRECTORY FOR SUBMISSION FILE
# ============================================================================
output_dir = os.getcwd()  # Root directory
artifacts_dir = os.path.join(output_dir, "rogii_artifacts")
if not os.path.exists(artifacts_dir):
    os.makedirs(artifacts_dir)

print("\n" + "="*80)
print("ROGII - WELLBORE GEOLOGY PREDICTION")
print(f"Submission saved to: {output_dir}")
print(f"Artifacts saved to: {artifacts_dir}")
print("="*80 + "\n")

# ============================================================================
# LOAD DATA
# ============================================================================
print("[STEP 1] LOADING DATA")
print("-" * 80)

try:
    train_df = pd.read_csv('../input/rogii-wellbore-geology-prediction/train.csv')
    test_df = pd.read_csv('../input/rogii-wellbore-geology-prediction/test.csv')
    print(f"✓ Train data: {train_df.shape}")
    print(f"✓ Test data: {test_df.shape}")
except FileNotFoundError:
    print("⚠ Competition data not found. Generating synthetic data...")
    np.random.seed(42)
    n_train = 5000
    n_test = 1000
    n_features = 20
    
    train_df = pd.DataFrame({
        f'feature_{i}': np.random.randn(n_train) for i in range(n_features)
    })
    train_df['id'] = [f'{i:08d}_{1442+j}' for i, j in enumerate(range(n_train))]
    train_df['tvt'] = np.random.uniform(0, 1, n_train)
    
    test_df = pd.DataFrame({
        f'feature_{i}': np.random.randn(n_test) for i in range(n_features)
    })
    test_df['id'] = [f'{n_train+i:08d}_{1442+j}' for i, j in enumerate(range(n_test))]
    
    print(f"✓ Synthetic train: {train_df.shape}")
    print(f"✓ Synthetic test: {test_df.shape}")

# ============================================================================
# DATA PREPROCESSING
# ============================================================================
print("\n[STEP 2] DATA PREPROCESSING")
print("-" * 80)

# Separate ID and target
train_ids = train_df['id'].copy()
test_ids = test_df['id'].copy()

X_train = train_df.drop(['id', 'tvt'], axis=1, errors='ignore')
y_train = train_df['tvt']
X_test = test_df.drop(['id'], axis=1, errors='ignore')

print(f"Features: {X_train.shape[1]}")
print(f"Train samples: {X_train.shape[0]}")
print(f"Test samples: {X_test.shape[0]}")

# Handle missing values
X_train = X_train.fillna(X_train.mean())
X_test = X_test.fillna(X_train.mean())

# Remove infinite values
mask = ~(np.isinf(X_train).any(axis=1) | np.isinf(y_train))
X_train = X_train[mask]
y_train = y_train[mask]
train_ids = train_ids[mask]

print(f"After cleaning: {X_train.shape[0]} samples")

# Scale features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
print("✓ Features scaled")

# ============================================================================
# MODEL TRAINING
# ============================================================================
print("\n[STEP 3] TRAINING MODELS")
print("-" * 80)

# Split for validation
X_tr, X_val, y_tr, y_val = train_test_split(X_train_scaled, y_train, test_size=0.2, random_state=42)

print("Training Linear Regression...")
lr_model = LinearRegression()
lr_model.fit(X_tr, y_tr)
lr_pred = lr_model.predict(X_val)
lr_rmse = np.sqrt(mean_squared_error(y_val, lr_pred))
print(f"  RMSE: {lr_rmse:.6f}")

print("Training Random Forest...")
rf_model = RandomForestRegressor(n_estimators=100, max_depth=15, random_state=42, n_jobs=-1)
rf_model.fit(X_tr, y_tr)
rf_pred = rf_model.predict(X_val)
rf_rmse = np.sqrt(mean_squared_error(y_val, rf_pred))
print(f"  RMSE: {rf_rmse:.6f}")

print("Training Gradient Boosting...")
gb_model = GradientBoostingRegressor(n_estimators=150, learning_rate=0.05, max_depth=5, 
                                      subsample=0.8, random_state=42, verbose=0)
gb_model.fit(X_tr, y_tr)
gb_pred = gb_model.predict(X_val)
gb_rmse = np.sqrt(mean_squared_error(y_val, gb_pred))
print(f"  RMSE: {gb_rmse:.6f}")

# Select best model
models = {'Linear Regression': (lr_model, lr_rmse), 
          'Random Forest': (rf_model, rf_rmse), 
          'Gradient Boosting': (gb_model, gb_rmse)}
best_model_name = min(models, key=lambda x: models[x][1])
best_model = models[best_model_name][0]
print(f"\n✓ Best Model: {best_model_name} (RMSE: {models[best_model_name][1]:.6f})")

# ============================================================================
# MAKE PREDICTIONS ON TEST SET
# ============================================================================
print("\n[STEP 4] GENERATING PREDICTIONS")
print("-" * 80)

test_predictions = best_model.predict(X_test_scaled)
print(f"Predictions generated: {len(test_predictions)}")
print(f"Min: {test_predictions.min():.6f}, Max: {test_predictions.max():.6f}, Mean: {test_predictions.mean():.6f}")

# ============================================================================
# CREATE SUBMISSION FILE IN ROOT DIRECTORY
# ============================================================================
print("\n[STEP 5] CREATING SUBMISSION FILE")
print("-" * 80)

submission = pd.DataFrame({
    'id': test_ids.values,
    'tvt': test_predictions
})

# SAVE TO ROOT DIRECTORY (not subdirectory)
submission_path = os.path.join(output_dir, 'submission.csv')
submission.to_csv(submission_path, index=False)
print(f"✓ submission.csv created in ROOT DIRECTORY")
print(f"   Path: {submission_path}")
print(f"   Rows: {len(submission)}")
print(f"\nFirst 5 rows:")
print(submission.head())

# Verify file exists
if os.path.exists(submission_path):
    print(f"\n✓ File verified: {os.path.getsize(submission_path)} bytes")
else:
    print(f"\n✗ ERROR: File not found at {submission_path}")

# ============================================================================
# SAVE MODELS & REPORTS (to artifacts folder)
# ============================================================================
print("\n[STEP 6] SAVING ARTIFACTS")
print("-" * 80)

# Save model
with open(os.path.join(artifacts_dir, 'best_model.pkl'), 'wb') as f:
    pickle.dump(best_model, f)
print("✓ best_model.pkl")

# Save scaler
with open(os.path.join(artifacts_dir, 'scaler.pkl'), 'wb') as f:
    pickle.dump(scaler, f)
print("✓ scaler.pkl")

# Save model info
model_info = {
    'best_model': best_model_name,
    'validation_rmse': float(models[best_model_name][1]),
    'all_models': {
        'Linear Regression': float(lr_rmse),
        'Random Forest': float(rf_rmse),
        'Gradient Boosting': float(gb_rmse)
    },
    'features': X_train.shape[1],
    'train_samples': X_train.shape[0],
    'test_samples': X_test.shape[0],
    'timestamp': datetime.now().strftime("%Y-%m-%d %H:%M:%S")
}

with open(os.path.join(artifacts_dir, 'model_info.json'), 'w') as f:
    json.dump(model_info, f, indent=2)
print("✓ model_info.json")

# Feature importance (if available)
if hasattr(best_model, 'feature_importances_'):
    feat_imp = pd.DataFrame({
        'feature': X_train.columns,
        'importance': best_model.feature_importances_
    }).sort_values('importance', ascending=False)
    feat_imp.to_csv(os.path.join(artifacts_dir, 'feature_importance.csv'), index=False)
    print("✓ feature_importance.csv")

# Save predictions with IDs for reference
pred_df = pd.DataFrame({
    'id': test_ids.values,
    'tvt_prediction': test_predictions
})
pred_df.to_csv(os.path.join(artifacts_dir, 'predictions_reference.csv'), index=False)
print("✓ predictions_reference.csv")

# ============================================================================
# VISUALIZATIONS (to artifacts folder)
# ============================================================================
print("\n[STEP 7] GENERATING VISUALIZATIONS")
print("-" * 80)

# Model comparison
fig, ax = plt.subplots(figsize=(10, 6))
model_names = list(models.keys())
rmse_scores = [models[m][1] for m in model_names]
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']
bars = ax.bar(model_names, rmse_scores, color=colors, edgecolor='black', linewidth=1.5)
ax.set_ylabel('RMSE', fontsize=12, fontweight='bold')
ax.set_title('Model Comparison - Validation RMSE', fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3, axis='y')
for bar, score in zip(bars, rmse_scores):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height,
            f'{score:.6f}', ha='center', va='bottom', fontweight='bold')
plt.tight_layout()
plt.savefig(os.path.join(artifacts_dir, '01_model_comparison.png'), dpi=300, bbox_inches='tight')
plt.close()
print("✓ 01_model_comparison.png")

# Prediction distribution
fig, ax = plt.subplots(figsize=(10, 6))
ax.hist(test_predictions, bins=50, color='steelblue', edgecolor='black', alpha=0.7)
ax.set_xlabel('Predicted TVT', fontsize=12, fontweight='bold')
ax.set_ylabel('Frequency', fontsize=12, fontweight='bold')
ax.set_title('Distribution of Test Predictions', fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.savefig(os.path.join(artifacts_dir, '02_prediction_distribution.png'), dpi=300, bbox_inches='tight')
plt.close()
print("✓ 02_prediction_distribution.png")

# Feature importance (if available)
if hasattr(best_model, 'feature_importances_'):
    fig, ax = plt.subplots(figsize=(10, 8))
    top_features = feat_imp.head(15)
    ax.barh(range(len(top_features)), top_features['importance'].values, color='coral', edgecolor='black')
    ax.set_yticks(range(len(top_features)))
    ax.set_yticklabels(top_features['feature'].values)
    ax.set_xlabel('Importance', fontsize=12, fontweight='bold')
    ax.set_title(f'Top 15 Features - {best_model_name}', fontsize=14, fontweight='bold')
    ax.grid(True, alpha=0.3, axis='x')
    plt.tight_layout()
    plt.savefig(os.path.join(artifacts_dir, '03_feature_importance.png'), dpi=300, bbox_inches='tight')
    plt.close()
    print("✓ 03_feature_importance.png")

# ============================================================================
# FINAL REPORT
# ============================================================================
print("\n" + "="*80)
print("✓ SUBMISSION READY!")
print("="*80)
print(f"\n📋 SUBMISSION FILE: submission.csv")
print(f"   Location: {submission_path}")
print(f"   Rows: {len(submission)}")
print(f"   Columns: id, tvt")
print(f"\n📊 ARTIFACTS FOLDER: {artifacts_dir}")
print(f"   ✓ best_model.pkl")
print(f"   ✓ scaler.pkl")
print(f"   ✓ model_info.json")
print(f"   ✓ feature_importance.csv")
print(f"   ✓ predictions_reference.csv")
print(f"   ✓ 3 PNG visualizations")
print(f"\n🏆 BEST MODEL: {best_model_name}")
print(f"   Validation RMSE: {models[best_model_name][1]:.6f}")
print("\n" + "="*80 + "\n")
```
