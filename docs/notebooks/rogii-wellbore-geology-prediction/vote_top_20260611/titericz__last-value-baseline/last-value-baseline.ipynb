```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
from glob import glob

# Read last TVT value in each test well
files = glob("/kaggle/input/competitions/rogii-wellbore-geology-prediction/test/*horizontal_well.csv")
MAP = {}
for fn in files:
    well_id = fn.split("/")[-1].split("_")[0]
    test = pd.read_csv(fn)
    MAP[well_id] = float(test.loc[test['TVT_input'].notnull(), 'TVT_input'].values[-1])

MAP
```

# Submist last value of each well

```python
submission = pd.read_csv("/kaggle/input/competitions/rogii-wellbore-geology-prediction/sample_submission.csv")
submission['well_id'] = submission['id'].apply(lambda x: x.split("_")[0])
submission['tvt'] = submission['well_id'].map(MAP)
submission['tvt'] = submission['tvt'].fillna(submission['tvt'].mean())
del submission['well_id']
submission.to_csv('submission.csv', index=False)
submission
```
